```python
import numpy as np

likes = np.array([120, 250, 180, 300, 150, 220, 275])
comments = np.array([10, 25, 15, 30, 12, 20, 18])

engagement = (likes + comments) / 7

print("Average engagement per post:", engagement)
```

```python
import numpy as np

prices = np.array([250, 300, 150, 400, 200])
discounts = np.array([20, 50, 30, 80, 25])

final_price = prices - discounts

print("Final prices:", final_price)
```

```python
import numpy as np

paytm = np.array([1, 0, 1, 0, 1, 0])
phonepe = np.array([0, 1, 1, 0, 0, 1])

result = np.logical_or(paytm, phonepe)

print("Paid by either app:", result)
```
