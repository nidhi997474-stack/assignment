```python
# Task 1 
import numpy as np

prices = np.array([299, 499, 799, 0, 1599, -1, 899])

print("Prices:", prices)
print("Data type:", prices.dtype)
```

```python
# Task 2
positive_prices = prices[prices > 0]
average_price = np.mean(positive_prices)

cleaned_prices = prices.copy()
cleaned_prices[cleaned_prices <= 0] = average_price

print("Average of positive prices:", average_price)
print("Cleaned prices:", cleaned_prices)
```

```python
# Task 3
quantities = np.array([2, 1, 3, 4, 2, 1, 5])

total_bill = cleaned_prices * quantities

print("Total bill for each item:", total_bill)
```

```python
# Task 4
print("Minimum price:", np.min(cleaned_prices))
print("Maximum price:", np.max(cleaned_prices))
print("Average price:", np.mean(cleaned_prices))
print("Total cleaned prices:", np.sum(cleaned_prices))
print("Total bill for all items:", np.sum(total_bill))
```

```python
# Task 5
unique_prices = np.unique(cleaned_prices)

print("Unique prices:", unique_prices)
print("Number of unique prices:", len(unique_prices))
```
