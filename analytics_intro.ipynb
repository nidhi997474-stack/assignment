```python
def total_likes(post):
    total = post["likes"]

    for reply in post["replies"]:
        total += total_likes(reply)

    return total


instagram_post = {
    "likes": 100,
    "replies": [
        {
            "likes": 50,
            "replies": [{
                    "likes": 20,
                    "replies": []}]},
        {
            "likes": 30,
            "replies": []}]}

print(total_likes(instagram_post))
```

```python
def insta_posts_generator(posts):
    for post in posts:
        yield post


posts = [
    "Enjoying the weekend!",
    "Coffee and coding ",
    "Beautiful sunset ",
    "Learning Python "
]

post_generator = insta_posts_generator(posts)

try:
    while True:
        print(next(post_generator))
except StopIteration:
    print("All posts printed.")
```

```python
def cashback_generator(transactions):
    for amount in transactions:
        yield amount * 0.05


transactions = [100, 250, 500, 1000]

cashback = cashback_generator(transactions)

for value in cashback:
    print("Cashback:", value)
```

```python
import re

def check_date(text):
    pattern = r'\b\d{2}/\d{2}/\d{4}\b'
    
    if re.search(pattern, text):
        return True
    else:
        return False

print(check_date("My appointment is on 25/06/2024"))
print(check_date("No date here"))
```

```python
import re

pattern = r'\bOD\d{18}\b'

orders = [
    "OD123456789012345000",
    "AB123456789012345000",
    "OD987654321098765432"
]

for order in orders:
    if re.search(pattern, order):
        print(order, "is valid")
    else:
        print(order, "is invalid")
```
