```python
# Task 1 
import numpy as np

zomato = np.array([4.5, 4.0, 3.5, 5.0, 4.2])
swiggy = np.array([4.0, 3.8, 4.3, 4.6, 3.9])

combined = np.concatenate((zomato, swiggy))

print("Combined ratings:", combined)
```

```python
# Task 2 
import numpy as np

post1 = np.array([100, 120, 150, 130, 140, 160, 180])
post2 = np.array([200, 220, 210, 250, 230, 240, 260])
post3 = np.array([80, 90, 100, 110, 95, 105, 120])

weekly_likes = np.vstack((post1, post2, post3))

print("Weekly likes:")
print(weekly_likes)

```

```python
# Task 4
import numpy as np

chat = np.array([
    [10, 12, 15, 20, 18, 22, 25],
    [5, 8, 7, 10, 9, 12, 11],
    [20, 18, 22, 25, 24, 27, 30]
])

new_user = np.array([[8, 10, 12, 9, 11, 13, 15]])

chat = np.insert(chat, 1, new_user, axis=0)

print("After adding new user:")
print(chat)

total_messages = np.sum(chat, axis=1)
least_user = np.argmin(total_messages)

chat = np.delete(chat, least_user, axis=0)

print("\nAfter removing user with least messages:")
print(chat)
```
