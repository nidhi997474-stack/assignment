```python

# Section C - Mini Capstone Project
# Food Delivery Analytics Console

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Generate synthetic dataset
np.random.seed(42)

n = 200

df = pd.DataFrame({
    "restaurant_name": np.random.choice(
        ["Spice Hub", "Burger House", "Pizza Point",
         "Food Corner", "Tasty Treats"], n
    ),
    "order_value": np.random.randint(100, 1500, n),
    "delivery_time_mins": np.random.randint(10, 70, n),
    "rating": np.round(np.random.uniform(1, 5, n), 1),
    "distance_km": np.round(np.random.uniform(1, 20, n), 2),
    "discount": np.random.randint(0, 200, n),
    "cuisine_type": np.random.choice(
        ["Indian", "Chinese", "Fast Food"], n
    )
})

# Check and handle missing numeric values
numeric_columns = df.select_dtypes(include=np.number).columns
df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].mean()
)

print("Dataset created successfully.")
print("Dataset shape:", df.shape)
print("Missing numeric values:")
print(df[numeric_columns].isnull().sum())


# 2. Menu Option 1 - Summary Statistics
def summary_statistics():
    print("\n--- Summary Statistics ---")
    print(df.describe().round(2))

    print("\nMean delivery time:",
          round(df["delivery_time_mins"].mean(), 2))

    print("Standard deviation of delivery time:",
          round(df["delivery_time_mins"].std(ddof=0), 2))


# 3. Menu Option 2 - Distribution Analysis
def distribution_analysis():
    print("\n--- Distribution Analysis ---")

    fig, axes = plt.subplots(1, 2, figsize=(12, 5))

    sns.histplot(
        data=df, x="order_value", kde=True, ax=axes[0]
    )
    axes[0].set_title("Order Value Distribution")

    sns.histplot(
        data=df, x="delivery_time_mins", kde=True, ax=axes[1]
    )
    axes[1].set_title("Delivery Time Distribution")

    fig.tight_layout()
    fig.savefig(
        "distribution_analysis.png",
        dpi=150,
        bbox_inches="tight"
    )
    plt.close(fig)

    print("Chart saved: distribution_analysis.png")


# 4. Menu Option 3 - Correlation Heatmap
def correlation_heatmap():
    print("\n--- Correlation Heatmap ---")

    corr = df.select_dtypes(include=np.number).corr(
        method="pearson"
    )

    fig, ax = plt.subplots(figsize=(9, 7))

    sns.heatmap(
        corr,
        annot=True,
        cmap="coolwarm",
        vmin=-1,
        vmax=1,
        fmt=".2f",
        ax=ax
    )

    ax.set_title("Food Delivery Correlation Heatmap")
    fig.tight_layout()
    fig.savefig(
        "correlation_heatmap.png",
        dpi=150,
        bbox_inches="tight"
    )
    plt.close(fig)

    print("Chart saved: correlation_heatmap.png")


# 5. Menu Option 4 - Restaurant Performance Report
def restaurant_performance_report():
    print("\n--- Restaurant Performance Report ---")

    report = df.groupby("restaurant_name").agg(
        mean_rating=("rating", "mean"),
        mean_order_value=("order_value", "mean"),
        mean_delivery_time=("delivery_time_mins", "mean")
    )

    report = report.sort_values(
        by="mean_rating",
        ascending=False
    )

    print(report.round(2))


# 6. Menu-Driven Console
while True:
    print("\n===== FOOD DELIVERY ANALYTICS CONSOLE =====")
    print("1. Summary Statistics")
    print("2. Distribution Analysis")
    print("3. Correlation Heatmap")
    print("4. Restaurant Performance Report")
    print("5. Exit")

    choice = input("Enter your choice (1-5): ").strip()

    if choice == "1":
        summary_statistics()

    elif choice == "2":
        distribution_analysis()

    elif choice == "3":
        correlation_heatmap()

    elif choice == "4":
        restaurant_performance_report()

    elif choice == "5":
        print("Exiting analytics console.")
        break

    else:
        print("Invalid choice. Please enter a number from 1 to 5.")


# 7. Final Summary Report - Runs After Exit
print("\n===== FINAL SUMMARY REPORT =====")

# Top 3 restaurants by mean rating
top_restaurants = (
    df.groupby("restaurant_name")["rating"]
    .mean()
    .sort_values(ascending=False)
    .head(3)
)

print("\nTop 3 Restaurants by Mean Rating:")
print(top_restaurants.round(2))

# Numeric pair with the highest absolute Pearson correlation
corr = df.select_dtypes(include=np.number).corr(
    method="pearson"
)

corr_values = corr.abs().where(
    np.triu(np.ones(corr.shape), k=1).astype(bool)
)

pair = corr_values.stack().idxmax()

print("\nHighest Absolute Pearson Correlation:")
print(
    pair[0], "and", pair[1],
    "=", round(corr.loc[pair[0], pair[1]], 3)
)

# Overall delivery-time statistics
print(
    "\nMean Delivery Time:",
    round(df["delivery_time_mins"].mean(), 2)
)

print(
    "Standard Deviation of Delivery Time:",
    round(df["delivery_time_mins"].std(ddof=0), 2)
)

print("\nAnalytics console completed successfully.")
```
