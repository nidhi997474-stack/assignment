```python
#  task 1 
import pandas as pd

# Create users DataFrame
users = pd.DataFrame({
    'user_id': [1, 2, 3, 4, 5],
    'username': ['Amit', 'Priya', 'Rahul', 'Neha', 'Karan'],
    'city': ['Ahmedabad', 'Mumbai', 'Delhi', 'Pune', 'Surat']
})

# Create orders DataFrame
orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105, 106, 107],
    'user_id': [1, 2, 1, 3, 5, 2, 7],
    'order_amount': [500, 750, 300, 900, 450, 600, 800]
})

# Inner join: show only users who placed orders
result = pd.merge(users, orders, on='user_id', how='inner')

print(result)
```

```python
#  task 2 
import pandas as pd

# Create users DataFrame
users = pd.DataFrame({
    'user_id': [1, 2, 3, 4, 5],
    'username': ['Amit', 'Priya', 'Rahul', 'Neha', 'Karan'],
    'city': ['Ahmedabad', 'Mumbai', 'Delhi', 'Pune', 'Surat']
})

# Create orders DataFrame
orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105, 106, 107],
    'user_id': [1, 2, 1, 3, 5, 2, 7],
    'order_amount': [500, 750, 300, 900, 450, 600, 800]
})

# Left join: keep all users, including those without orders
result = pd.merge(users, orders, on='user_id', how='left')

print(result)
```

```python
#  task 3 
import pandas as pd

# Create restaurants DataFrame
restaurants = pd.DataFrame({
    'restaurant_id': [1, 2, 3, 4],
    'name': ['Spice Garden', 'Pizza House', 'Tasty Bites', 'Green Bowl'],
    'city': ['Ahmedabad', 'Mumbai', 'Delhi', 'Pune']
})

# Create reviews DataFrame
zomato_reviews = pd.DataFrame({
    'review_id': [201, 202, 203, 204],
    'restaurant_id': [1, 2, 5, 2],
    'rating': [4.5, 4.0, 3.5, 5.0],
    'reviewer_city': ['Ahmedabad', 'Mumbai', 'Surat', 'Pune']
})

# Outer join: include all restaurants and all reviews
result = pd.merge(
    restaurants,
    zomato_reviews,
    on='restaurant_id',
    how='outer'
)

print(result)
```

```python
#  task 4
import pandas as pd

# Create playlists DataFrame
playlists = pd.DataFrame({
    'playlist_id': [101, 102, 103, 104, 105],
    'user_id': [1, 2, 3, 4, 5],
    'city': ['Ahmedabad', 'Mumbai', 'Delhi', 'Pune', 'Surat'],
    'genre': ['Pop', 'Rock', 'Classical', 'Jazz', 'Hip-Hop']
})

# Create user profiles DataFrame
user_profiles = pd.DataFrame({
    'user_id': [1, 2, 3, 4, 5],
    'username': ['Amit', 'Priya', 'Rahul', 'Neha', 'Karan'],
    'city': ['Ahmedabad', 'Mumbai', 'Jaipur', 'Pune', 'Surat']
})

# Merge on both user_id and city
result = pd.merge(
    playlists,
    user_profiles,
    on=['user_id', 'city'],
    how='inner'
)

print(result)
```

```python
#  task 5 
import pandas as pd

# Create Flipkart products DataFrame
products = pd.DataFrame({
    'product_id': [101, 102, 103, 104],
    'name': ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
    'category': ['Electronics', 'Accessories', 'Accessories', 'Electronics']
})

# Create product ratings DataFrame
ratings = pd.DataFrame({
    'rating_id': [1, 2, 3, 4, 5],
    'product_id': [101, 102, 102, 105, 103],
    'rating': [4.5, 4.0, 3.5, 5.0, 4.2]
})

# Right join: keep all ratings, even without a matching product
result = pd.merge(
    products,
    ratings,
    on='product_id',
    how='right'
)

print(result)
```
