```python
#  task 3
import pandas as pd

# Create Zomato food orders dataset
food_orders = pd.DataFrame({
    'restaurant': [
        'Pizza Hut', 'Burger King', 'Dominos',
        'Biryani House', 'Subway'
    ],
    'order_amount': [750, 350, 620, 900, 450],
    'delivery_time': [35, 25, 40, 45, 20]
})

# Filter orders where order_amount is greater than 500
filtered_orders = food_orders[food_orders['order_amount'] > 500]

print(filtered_orders)
```

```python
# task 4 
import pandas as pd

# Create Spotify songs dataset
songs = pd.DataFrame({
    'song': ['Song A', 'Song B', 'Song C', 'Song D', 'Song E'],
    'artist': [
        'Artist A', 'Artist B', 'Artist C', 'Artist D', 'Artist E'
    ],
    'streams': [1500000, 800000, 2000000, 1200000, 3000000],
    'duration': [170, 160, 200, 150, 175]
})

# Select songs with over 1,000,000 streams and duration below 180 seconds
result = songs.query('streams > 1000000 and duration < 180')

print(result)
```

```python
# task 5 
import pandas as pd

# Create IPL cricket matches dataset
matches = pd.DataFrame({
    'team1': ['CSK', 'MI', 'RCB', 'KKR', 'RR'],
    'team2': ['MI', 'RCB', 'KKR', 'RR', 'CSK'],
    'winner': ['CSK', 'MI', 'RCB', 'KKR', 'RR'],
    'total_runs': [190, 175, 220, 205, 150]
})

# Filter matches with total_runs between 180 and 220, inclusive
result = matches[
    (matches['total_runs'] >= 180) &
    (matches['total_runs'] <= 220)
]

print(result)
```
