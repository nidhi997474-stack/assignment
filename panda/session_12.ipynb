```python
#  taks 1 
import pandas as pd

# Create sample IPL match data
matches = pd.DataFrame({
    'team': ['CSK', 'MI', 'RCB', 'CSK', 'MI', 'RCB', 'CSK'],
    'runs': [180, 165, 200, 195, 175, 190, 160]
})

# Calculate total runs for each team
team_runs = matches.groupby('team')['runs'].sum().reset_index()

print(team_runs)
```

```python
#  task 2 
import pandas as pd

# Create Zomato food orders
orders = pd.DataFrame({
    'city': ['Ahmedabad', 'Ahmedabad', 'Mumbai', 'Mumbai',
             'Ahmedabad', 'Mumbai'],
    'restaurant': ['Spice Garden', 'Pizza House', 'Spice Garden',
                   'Pizza House', 'Spice Garden', 'Pizza House'],
    'order_amount': [500, 700, 600, 800, 900, 400]
})

# Calculate average order amount by city and restaurant
average_orders = orders.groupby(
    ['city', 'restaurant']
)['order_amount'].agg('mean').reset_index()

print(average_orders)
```

```python
#  task 3 
import pandas as pd

# Create Flipkart-style product sales data
sales = pd.DataFrame({
    'category': ['Electronics', 'Fashion', 'Electronics',
                 'Fashion', 'Electronics', 'Fashion'],
    'region': ['West', 'West', 'East', 'East', 'West', 'West'],
    'sales': [50000, 12000, 40000, 15000, 30000, 18000],
    'units_sold': [10, 20, 8, 25, 6, 30]
})

# Apply multiple aggregations
result = sales.groupby(['category', 'region']).agg(
    total_sales=('sales', 'sum'),
    average_units_sold=('units_sold', 'mean')
).reset_index()

print(result)
```

```python
#  task 4 
import pandas as pd

# Create regional revenue data
sales = pd.DataFrame({
    'region': ['West', 'East', 'North', 'West', 'East', 'North', 'West'],
    'revenue': [50000, 40000, 30000, 70000, 60000, 45000, 65000]
})

# Calculate multiple aggregations using a dictionary
result = sales.groupby('region').agg({
    'revenue': ['sum', 'max', 'min']
})

# Rename columns for clarity
result.columns = ['total_revenue', 'max_revenue', 'min_revenue']

print(result)
```

```python
#  task 5 
import pandas as pd

# Create Spotify song streams data
songs = pd.DataFrame({
    'artist': ['Artist A', 'Artist B', 'Artist C', 'Artist D',
               'Artist E', 'Artist F', 'Artist G', 'Artist H'],
    'genre': ['Pop', 'Rock', 'Pop', 'Jazz', 'Rock', 'Jazz', 'Pop', 'Rock'],
    'streams': [1000000, 800000, 1500000, 400000, 1200000, 600000, 900000, 700000]
})

# Calculate total and average streams by genre
result = songs.groupby('genre').agg(
    total=('streams', 'sum'),
    average=('streams', 'mean')
)

# Sort by total streams in descending order
result = result.sort_values(by='total', ascending=False)

print(result)
```
