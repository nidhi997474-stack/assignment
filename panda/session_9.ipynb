```python
#  task 1 
import pandas as pd

# Create a Series of date strings in DD-MM-YYYY format
dates = pd.Series([
    '12-01-2026',
    '25-03-2026',
    '07-05-2026',
    '18-08-2026',
    '30-11-2026'
])

# Convert strings to datetime objects
dates = pd.to_datetime(dates, format='%d-%m-%Y')

print(dates)
```

```python
#  task 2 
import pandas as pd

# Create movie dataset
movies = pd.DataFrame({
    'title': [
        '3 Idiots',
        'Dangal',
        'Jawan',
        'Pathaan',
        'PK'
    ],
    'release_date': [
        '25-12-2009',
        '23-12-2016',
        '07-09-2023',
        '25-01-2023',
        '19-12-2014'
    ]
})

# Convert release_date to datetime
movies['release_date'] = pd.to_datetime(
    movies['release_date'], format='%d-%m-%Y'
)
# Extract year, month, and day
movies['year'] = movies['release_date'].dt.year
movies['month'] = movies['release_date'].dt.month
movies['day'] = movies['release_date'].dt.day

print(movies)
```

```python
#  task 3 
import pandas as pd

# Generate today's date and the following 6 days
date_range = pd.date_range(
    start=pd.Timestamp.today().normalize(),
    periods=7,
    freq='D'
)

print("Next 7 days:")
print(date_range)
```

```python
#  task 4 
import pandas as pd

# Create Zomato Gold subscription dataset
subscriptions = pd.DataFrame({
    'user_id': [101, 102, 103, 104, 105],
    'start_date': [
        '01-01-2026',
        '15-03-2026',
        '01-06-2026',
        '20-08-2026',
        '01-10-2026'
    ]
})

# Convert start_date to datetime
subscriptions['start_date'] = pd.to_datetime(
    subscriptions['start_date'], format='%d-%m-%Y'
)

# Calculate tenure in days from today's date
today = pd.Timestamp.today().normalize()

subscriptions['tenure_days'] = (
    today - subscriptions['start_date']
).dt.days

print(subscriptions)
```
