```python
# task 3 
import pandas as pd

followers = pd.Series([1200, 54000, 780, 150000, 32000])

print("Followers:")
print(followers)

print("\nHighest follower count:", followers.max())
print("Lowest follower count:", followers.min())

print("\nInfluencer with highest followers:")
print(followers.idxmax())

print("\nInfluencer with lowest followers:")
print(followers.idxmin())
```
