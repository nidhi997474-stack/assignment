```python
# task 1 
import pandas as pd

# Create sample Flipkart product sales data
sales = pd.DataFrame({
    'ProductName': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Headphones'],
    'Price': [55000, 500, 1200, 12000, 2000],
    'Qty': [2, 5, 3, 1, 4]
})

# Add TotalValue column
sales['TotalValue'] = sales['Price'] * sales['Qty']

print(sales)
```

```python
#  task 2 
import pandas as pd

# Create food items dataset
food = pd.DataFrame({
    'Item': ['Pizza', 'Burger', 'Sandwich', 'Pasta', 'Fries'],
    'Price': [300, 150, 120, 250, 100],
    'Qty': [2, 3, 1, 2, 4]
})

# Apply 10% discount to each item's price
food['DiscountedPrice'] = food['Price'].apply(lambda price: price * 0.90)

print(food)
```

```python
#  task 3 
import pandas as pd

# Create Instagram posts dataset
posts = pd.DataFrame({
    'PostID': [101, 102, 103, 104, 105],
    'Likes': [1500, 2000000, 850, 1200000, 45000],
    'Comments': [1200, 25000, 950, 2000000, 15000]
})

# Helper function to format large numbers
def format_number(number):
    if number >= 1000000:
        value = number / 1000000
        return f'{value:g}M'
    elif number >= 1000:
        value = number / 1000
        return f'{value:g}K'
    else:
        return str(number)

# Format Likes and Comments as strings
posts[['Likes', 'Comments']] = posts[['Likes', 'Comments']].apply(
    lambda column: column.map(format_number)
)

print(posts)
```

```python
#  task 4 
import pandas as pd

# Create Zomato restaurant orders
orders = pd.DataFrame({
    'Restaurant': ['Dominos', 'Pizza Hut', 'Subway', 'Burger King', 'Biryani House'],
    'Price': [300, 450, 200, 250, 350],
    'Qty': [2, 1, 3, 2, 2],
    'DeliveryCharge': [40, 30, 25, 35, 20]
})

# Calculate final amount = (Price * Qty) + DeliveryCharge
orders['FinalAmount'] = orders.apply(
    lambda row: (row['Price'] * row['Qty']) + row['DeliveryCharge'],
    axis=1
)

print(orders)
```

```python
#  task 5 
import pandas as pd

# Create Myntra orders dataset
orders = pd.DataFrame({
    'OrderID': [101, 102, 103, 104, 105],
    'ProductName': ['Shoes', 'T-Shirt', 'Jeans', 'Jacket', 'Watch'],
    'Price': [2500, 800, 1800, 3500, 1200],
    'Qty': [1, 3, 2, 1, 2]
})

# Calculate GST as 5% of Price * Qty
orders['GST'] = (orders['Price'] * orders['Qty']) * 0.05

print(orders)
```
