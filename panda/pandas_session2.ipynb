```python
# task 3 
import pandas as pd

# Sample food delivery dataset
food_data = pd.DataFrame({
    'Restaurant': ['Pizza Hub', 'Burger Point', 'Spice House',
                   'Pizza Hub', 'Spice House', 'Burger Point'],
    'Rating': [4.5, 4.0, 4.2, 4.8, 3.9, 4.1],
    'Orders': [120, 90, 150, 110, 80, 100],
    'Average_Bill': [450, 250, 350, 500, 300, 275]
})

# Display summary statistics
print("Summary Statistics:")
print(food_data.describe())

# Display DataFrame information
print("\nDataFrame Information:")
food_data.info()

# Two insights
print("\nInsights:")
print("1. The average restaurant rating is approximately",
      round(food_data['Rating'].mean(), 2))
print("2. The highest number of orders in a record is",
      food_data['Orders'].max())
```
