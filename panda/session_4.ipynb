```python
# task 1 
import pandas as pd

movies = pd.DataFrame({
    'title': ['3 Idiots', 'Dangal', 'Jawan', 'PK', 'Lagaan', 'Taare Zameen Par'],
    'rating': [8.4, 8.3, 7.0, 8.1, 8.1, 8.3],
    'release_year': [2009, 2016, 2023, 2014, 2001, 2007]
})

# Sort ratings from highest to lowest
sorted_movies = movies.sort_values(by='rating', ascending=False)

print(sorted_movies)
```

```python
#  task 2
import pandas as pd

teams = pd.DataFrame({
    'team': ['CSK', 'MI', 'RCB', 'KKR', 'RR'],
    'points': [17, 16, 14, 12, 10],
    'net_run_rate': [0.8, 0.6, 0.3, 0.1, -0.2]
})

# Sort the index in descending order
sorted_teams = teams.sort_index(ascending=False)

print(sorted_teams)
```

```python
# task 3 
import pandas as pd

videos = pd.DataFrame({
    'video_title': [
        'Python Tutorial', 'Music Video', 'Cricket Highlights',
        'Travel Vlog', 'Cooking Recipe'
    ],
    'views': [150000, 500000, 300000, 120000, 90000],
    'likes': [12000, 45000, 25000, 8000, 6000]
}, index=[10, 20, 30, 40, 50])

# Reset index without adding the old index as a column
videos = videos.reset_index(drop=True)

print(videos)
```

```python
# task 4 
import pandas as pd

restaurants = pd.DataFrame({
    'restaurant': ['Dominos', 'Burger King', 'Biryani House', 'Pizza Hut'],
    'rating': [4.2, 4.0, 4.5, 4.3]
}, index=['Dominos', 'Burger King', 'Biryani House', 'Pizza Hut'])

# Put your favorite restaurant first
restaurants = restaurants.reindex([
    'Biryani House', 'Dominos', 'Burger King', 'Pizza Hut'
])

print(restaurants)
```

```python
#  task 5
import pandas as pd

products = pd.DataFrame({
    'product_name': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Headphones'],
    'price': [55000, 500, 1200, 12000, 2000],
    'discount': [10, 30, 20, 15, 40]
})

# Sort by discount from highest to lowest
products = products.sort_values(by='discount', ascending=False)

# Reset index and discard the old index
products = products.reset_index(drop=True)

print(products)
```
