```python
#  task 1 
import pandas as pd

# Create sample IPL batsmen dataset
ipl = pd.DataFrame({
    'player_name': ['Virat Kohli', 'Shubman Gill', 'Faf du Plessis',
                    'David Warner', 'Ruturaj Gaikwad', 'Sanju Samson',
                    'Suryakumar Yadav', 'KL Rahul', 'Ishan Kishan', 'Rinku Singh'],
    'runs': [639, 890, 730, 516, 590, 362, 605, 274, 454, 474],
    'matches': [14, 17, 14, 14, 16, 14, 16, 9, 16, 14],
    'average': [53.25, 59.33, 56.15, 36.86, 42.14, 30.17, 43.21, 34.25, 30.27, 59.25],
    'strike_rate': [139.82, 157.80, 153.68, 131.63, 147.50, 153.39, 181.14, 113.22, 142.77, 149.52]
})

# Display first 10 rows
print(ipl.head(10))
```

```python
#  task 2
# Calculate summary statistics for runs
print("Mean:", ipl['runs'].mean())
print("Median:", ipl['runs'].median())
print("Mode:", ipl['runs'].mode().tolist())
print("Minimum:", ipl['runs'].min())
print("Maximum:", ipl['runs'].max())
```

```python
#  task 3 
import pandas as pd

# Create sample restaurant data
restaurants = pd.DataFrame({
    'name': ['Restaurant A', 'Restaurant B', 'Restaurant C',
             'Restaurant D', 'Restaurant E'],
    'rating': [4.5, 4.2, 3.9, 4.7, 4.0],
    'votes': [1200, 850, 400, 1800, 600],
    'avg_cost': [800, 500, 350, 1200, 450]
})

# Calculate correlation between rating and votes
correlation = restaurants['rating'].corr(restaurants['votes'])

print(restaurants)
print("\nCorrelation between rating and votes:", round(correlation, 3))
```

```python
#  task 4 
# Generate summary statistics for numeric columns
summary = ipl.describe()

print(summary)

# Identify the batsman with the highest runs
highest_scorer = ipl.loc[ipl['runs'].idxmax()]

print("\nHighest-scoring batsman:")
print(highest_scorer[['player_name', 'runs']])
```

```python
# task 5
#  Correlation between matches played and runs
correlation = ipl['matches'].corr(ipl['runs'])

print("Correlation between matches and runs:", round(correlation, 3))

# Display the five highest run scorers
top_scorers = ipl.nlargest(5, 'runs')[['player_name', 'matches', 'runs']]

print("\nTop 5 run scorers:")
print(top_scorers)
```
