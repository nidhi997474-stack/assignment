```python
#  task 1
import pandas as pd

# Create IPL players dataset
players = pd.DataFrame({
    'player_name': ['Virat Kohli', 'MS Dhoni', 'Rohit Sharma',
                    'Jasprit Bumrah', 'Ravindra Jadeja'],
    'team': ['RCB', 'CSK', 'MI', 'MI', 'CSK'],
    'age': [37, None, 39, None, 37]
})

# Save and load CSV
players.to_csv('ipl_players.csv', index=False)
players = pd.read_csv('ipl_players.csv')

# Display rows where age is missing
print(players[players['age'].isnull()])
```

```python
#  task 2 
import pandas as pd

# Create food delivery dataset
food_orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105],
    'customer_name': ['Amit', 'Priya', 'Rahul', 'Neha', 'Karan'],
    'delivery_rating': [4.5, None, 3.8, None, 4.9]
})

# Remove rows with missing delivery_rating
cleaned_orders = food_orders.dropna(subset=['delivery_rating'])

print(cleaned_orders)
```

```python
# task 3 
import pandas as pd

# Create Flipkart-style product reviews
products = pd.DataFrame({
    'product_name': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Headphones'],
    'rating': [4.5, 4.0, None, 3.5, None]
})

# Calculate mean of available ratings
mean_rating = products['rating'].mean()

# Fill missing ratings with the mean
products['rating'] = products['rating'].fillna(mean_rating)

print("Mean rating:", mean_rating)
print(products)
```

```python
#  task 4 
import pandas as pd

# Create Spotify playlist dataset
playlist = pd.DataFrame({
    'song': ['Song A', 'Song B', 'Song C', 'Song D', 'Song E'],
    'artist': ['Artist A', 'Artist B', 'Artist C', 'Artist D', 'Artist E'],
    'duration': [180, 210, None, 240, None]
})

# Calculate median of available durations
median_duration = playlist['duration'].median()

# Replace missing durations with the median
playlist['duration'] = playlist['duration'].fillna(median_duration)

# Save the cleaned dataset
playlist.to_csv('cleaned_playlist.csv', index=False)

print("Median duration:", median_duration)
print(playlist)
print("\nSaved as cleaned_playlist.csv")
```

```python
# task 5 
import pandas as pd

# Create sample Instagram user data
followers = pd.DataFrame({
    'username': ['user1', 'user2', 'user3', 'user4', 'user5'],
    'followers': [1200, None, 5400, None, 3200]
})

# Replace all missing follower counts with zero
followers['followers'] = followers['followers'].fillna(0)

print("Updated Instagram data:")
print(followers)

# Verify that no missing values remain in followers
print("\nMissing values remaining:", followers['followers'].isnull().sum())
```
