```python
#  task 2 
import pandas as pd

reviews = pd.DataFrame({
    'product_id': [101, 101, 102, 103, 101, 102],
    'user_id': [1, 1, 2, 3, 1, 2],
    'rating': [5, 5, 4, 3, 5, 4],
    'review_text': ['Excellent', 'Excellent', 'Good', 'Average',
                    'Excellent', 'Good']
})

# Count duplicate reviews before removal
duplicates_before = reviews.duplicated(
    subset=['product_id', 'user_id']
).sum()

# Remove duplicates, keeping the first occurrence
cleaned_reviews = reviews.drop_duplicates(
    subset=['product_id', 'user_id'],
    keep='first'
)

# Count duplicate reviews after removal
duplicates_after = cleaned_reviews.duplicated(
    subset=['product_id', 'user_id']
).sum()

print("Duplicate reviews before removal:", duplicates_before)
print("Duplicate reviews after removal:", duplicates_after)
print("\nCleaned DataFrame:")
print(cleaned_reviews)
```

```python
# task 3 
import pandas as pd

orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105, 106, 107, 108],
    'order_amount': [250, 300, 350, 400, 450, 500, 550, 2000]
})

# Calculate Q1, Q3, and IQR
Q1 = orders['order_amount'].quantile(0.25)
Q3 = orders['order_amount'].quantile(0.75)
IQR = Q3 - Q1

# Calculate outlier limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Identify outliers
outliers = orders[
    (orders['order_amount'] < lower_limit) |
    (orders['order_amount'] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("\nOutlier orders:")
print(outliers[['order_id', 'order_amount']])
```

```python
import pandas as pd

# Create daily IPL ticket-sales data
ticket_sales = pd.DataFrame({
    'date': [
        '2026-10-01', '2026-10-02', '2026-10-03',
        '2026-10-04', '2026-10-05', '2026-10-06',
        '2026-10-07'
    ],
    'tickets_sold': [100, 110, 105, 95, 120, 108, 1000]
})

# Calculate Z-scores
ticket_sales['z_score'] = (
    ticket_sales['tickets_sold'] - ticket_sales['tickets_sold'].mean()
) / ticket_sales['tickets_sold'].std(ddof=0)

# Select dates where absolute Z-score is greater than 2
outliers = ticket_sales[ticket_sales['z_score'].abs() > 2]

print("Ticket sales with Z-scores:")
print(ticket_sales)

print("\nOutlier dates:")
print(outliers[['date', 'tickets_sold', 'z_score']])
```

```python
#  task 5 
import pandas as pd
def find_outliers(transactions):
    mean_amount = transactions['amount'].mean()
    std_amount = transactions['amount'].std(ddof=0)

    if std_amount == 0:
        return transactions.iloc[0:0].copy()

    z_scores = (transactions['amount'] - mean_amount) / std_amount

    return transactions[z_scores.abs() > 2]

# Create sample Paytm transaction data
transactions = pd.DataFrame({
    'transaction_id': [1, 2, 3, 4, 5, 6, 7],
    'amount': [100, 150, 120, 130, 110, 140, 5000]
})

# Test the function
result = find_outliers(transactions)

print("All transactions:")
print(transactions)

print("\nOutlier transactions:")
print(result)
```
