```python
# SESSION 14 - FULL PANDAS MINI PROJECT

import pandas as pd
import numpy as np

# TASK 1: Load IPL match data into a DataFrame
# Sample data included so the code runs without an external CSV file
ipl_df = pd.DataFrame({
    'season': [2023, 2023, 2023, 2023, 2023, 2022, 2022, 2022, 2022, 2022],
    'team1': [
        'CSK', 'MI', 'RCB', 'GT', 'RR',
        'CSK', 'MI', 'RCB', 'GT', 'RR'
    ],
    'team2': [
        'MI', 'RCB', 'GT', 'RR', 'CSK',
        'RCB', 'GT', 'RR', 'CSK', 'MI'
    ],
    'winner': [
        'CSK', 'MI', 'GT', 'GT', 'CSK',
        'CSK', 'GT', 'RCB', 'GT', 'RR'
    ],
    'win_by_runs': [7, 0, 0, 6, 15, 13, 0, 0, 0, 0],
    'win_by_wickets': [0, 5, 4, 0, 0, 0, 6, 3, 7, 4]
})

print("TASK 1: First 5 rows")
print(ipl_df.head(5))


# TASK 2: Remove rows with missing winners and reset index
ipl_df = ipl_df.dropna(subset=['winner']).reset_index(drop=True)

print("\nTASK 2: Cleaned IPL DataFrame")
print(ipl_df)


# TASK 3: Create match margin type
ipl_df['match_margin_type'] = np.select(
    [
        ipl_df['win_by_runs'] > 0,
        ipl_df['win_by_wickets'] > 0
    ],
    [
        'Runs',
        'Wickets'
    ],
    default='Tie/No Result'
)

print("\nTASK 3: Match margin type")
print(ipl_df[['team1', 'team2', 'winner', 'match_margin_type']])


# TASK 4: Create team home-city data and merge
team_cities = pd.DataFrame({
    'team': ['CSK', 'MI', 'RCB', 'GT', 'RR'],
    'home_city': [
        'Chennai',
        'Mumbai',
        'Bengaluru',
        'Ahmedabad',
        'Jaipur'
    ]
})

# Merge home cities for team1
ipl_df = pd.merge(
    ipl_df,
    team_cities.rename(columns={
        'team': 'team1',
        'home_city': 'team1_home_city'
    }),
    on='team1',
    how='left'
)

# Merge home cities for team2
ipl_df = pd.merge(
    ipl_df,
    team_cities.rename(columns={
        'team': 'team2',
        'home_city': 'team2_home_city'
    }),
    on='team2',
    how='left'
)

print("\nTASK 4: Teams with their home cities")
print(ipl_df[[
    'team1', 'team1_home_city',
    'team2', 'team2_home_city'
]])


# TASK 5: Pivot table showing wins by team per season
wins_pivot = pd.pivot_table(
    ipl_df,
    index='winner',
    columns='season',
    values='team1',
    aggfunc='count',
    fill_value=0
)

print("\nTASK 5: Total matches won by each team per season")
print(wins_pivot)

# Find the top 3 teams by wins in any single season
top_teams = (
    wins_pivot
    .max(axis=1)
    .sort_values(ascending=False)
    .head(3)
)

print("\nTop 3 teams with the most wins in a single season")
print(top_teams)
```
