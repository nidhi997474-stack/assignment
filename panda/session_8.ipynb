```python
# task 1 
import pandas as pd

# Create Instagram usernames
users = pd.DataFrame({
    'username': [
        'CoolDude',
        'foodieQueen',
        'Tech_Girl',
        'TravelLover',
        'MusicFan'
    ]
})

# Convert usernames to lowercase
users['username'] = users['username'].str.lower()

print(users)
```

```python
#  task 2 
import pandas as pd

# Create Spotify song titles
songs = pd.Series([
    'Perfect Remix',
    'Kesariya Remix',
    'Shape of You',
    'Tum Hi Ho Remix',
    'Blinding Lights Remix'
])

# Remove Remix from song titles
cleaned_songs = songs.str.replace('Remix', '', regex=False).str.strip()

print(cleaned_songs)
```

```python
#  task 3 
import pandas as pd

# Create Zomato customer reviews
reviews = pd.DataFrame({
    'review': [
        'Fast delivery and good food',
        'The food was delicious',
        'Late Delivery but tasty food',
        'Excellent packaging',
        'Quick home delivery'
    ]
})

# Select reviews containing delivery, ignoring case
filtered_reviews = reviews[
    reviews['review'].str.contains('delivery', case=False, na=False)
]

print(filtered_reviews)
```

```python
#  task 4 
import pandas as pd

# Create Flipkart product categories
products = pd.DataFrame({
    'category': [
        'Electronics/Mobiles/Smartphones',
        'Home/Kitchen/Appliances',
        'Fashion/Men/Shirts',
        'Electronics/Computers/Laptops',
        'Home/Furniture/Chairs'
    ]
})

# Extract the main category before the first /
products['main_category'] = products['category'].str.split('/').str[0]

print(products)
```

```python
#  task 5 
import pandas as pd

# Create YouTube video titles
videos = pd.DataFrame({
    'title': [
        'Python Pandas Tutorial',
        'How to Learn Python',
        'Data Analysis Basics',
        'Excel Tips and Tricks',
        'Machine Learning Guide'
    ]
})

# Convert to uppercase and replace spaces with underscores in one line
videos['title'] = videos['title'].str.upper().str.replace(' ', '_', regex=False)

print(videos)
```
