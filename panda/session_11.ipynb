```python
#  task 1 
import pandas as pd

# January orders
january_orders = pd.DataFrame({
    'order_id': [101, 102, 103],
    'customer': ['Amit', 'Priya', 'Rahul'],
    'item': ['Pizza', 'Burger', 'Pasta'],
    'amount': [350, 200, 450]
})

# February orders
february_orders = pd.DataFrame({
    'order_id': [104, 105, 106],
    'customer': ['Neha', 'Karan', 'Riya'],
    'item': ['Sandwich', 'Biryani', 'Pizza'],
    'amount': [150, 300, 400]
})

# Combine row-wise
all_orders = pd.concat(
    [january_orders, february_orders],
    ignore_index=True
)

print(all_orders)
```

```python
#  task 2 
import pandas as pd

# Product details
products = pd.DataFrame({
    'product_id': [101, 102, 103, 104, 105],
    'product_name': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Headphones'],
    'price': [55000, 500, 1200, 12000, 2000]
})

# Corresponding product ratings in the same row order
ratings = pd.DataFrame({
    'rating': [4.5, 4.0, 4.2, 4.6, 4.1]
})

# Combine column-wise
products_with_ratings = pd.concat(
    [products, ratings],
    axis=1
)

print(products_with_ratings)
```

```python
#  task 3 
import pandas as pd

# Daily streams for three users over one week
streams = pd.DataFrame({
    'user': [
        'Amit', 'Amit', 'Amit', 'Amit', 'Amit', 'Amit', 'Amit',
        'Priya', 'Priya', 'Priya', 'Priya', 'Priya', 'Priya', 'Priya',
        'Rahul', 'Rahul', 'Rahul', 'Rahul', 'Rahul', 'Rahul', 'Rahul'
    ],
    'day': [
        'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'
    ] * 3,
    'streams': [
        120, 150, 130, 170, 200, 250, 220,
        100, 140, 160, 180, 210, 230, 190,
        90, 110, 130, 150, 170, 200, 240
    ]
})

# Reshape: each row is a user and each column is a day
weekly_streams = streams.pivot(
    index='user',
    columns='day',
    values='streams'
)

print(weekly_streams)
```

```python
#  task 4 
import pandas as pd

# Zomato restaurant orders
orders = pd.DataFrame({
    'restaurant': [
        'Spice Garden', 'Spice Garden', 'Spice Garden',
        'Pizza House', 'Pizza House', 'Pizza House',
        'Tasty Bites', 'Tasty Bites', 'Tasty Bites'
    ],
    'month': [
        'January', 'February', 'March',
        'January', 'February', 'March',
        'January', 'February', 'March'
    ],
    'total_amount': [500, 700, 600, 800, 900, 750, 300, 450, 400]
})

# Average order amount per restaurant for each month
monthly_average = pd.pivot_table(
    orders,
    index='restaurant',
    columns='month',
    values='total_amount',
    aggfunc='mean'
)

print(monthly_average)
```

```python
#  task 5 
import pandas as pd

# January sales
sales_jan = pd.DataFrame({
    'product': ['Shoes', 'T-Shirt', 'Jeans'],
    'sales': [10000, 5000, 8000]
})

# February sales
sales_feb = pd.DataFrame({
    'product': ['Shoes', 'T-Shirt', 'Jeans'],
    'sales': [12000, 7000, 6000]
})

# March sales
sales_mar = pd.DataFrame({
    'product': ['Shoes', 'T-Shirt', 'Jeans'],
    'sales': [15000, 6000, 9000]
})

# Combine monthly sales
all_sales = pd.concat(
    [sales_jan, sales_feb, sales_mar],
    ignore_index=True
)

# Calculate total sales per product
total_sales = all_sales.groupby('product', as_index=False)['sales'].sum()

print("Combined monthly sales:")
print(all_sales)

print("\nTotal sales per product:")
print(total_sales)
```
