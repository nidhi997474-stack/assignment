```python
# Task 1 
import numpy as np

teams = np.array([
    "MI", "CSK", "RCB", "MI", "KKR",
    "CSK", "RCB", "MI", "KKR", "CSK"
])

unique_teams = np.unique(teams)

print("Unique team names:", unique_teams)
```

```python
# Task 2 
import numpy as np

ratings = np.array([4.5, np.nan, 3.8, 4.2, np.nan, 5.0])

missing = np.isnan(ratings)

print("Ratings:", ratings)
print("Number of missing ratings:", np.sum(missing))
```

```python
# Task 3
import numpy as np

prices = np.array([50, 250, 1200, 750, 90, 1500, 500])

limited_prices = np.clip(prices, 100, 1000)

print("Original prices:", prices)
print("Clipped prices:", limited_prices)
```

```python
# Task 4
import numpy as np

views = np.array([1000, np.nan, 5000, np.inf, 3000, 8000, np.nan])

finite_values = views[np.isfinite(views)]
max_finite = np.max(finite_values)

result = np.nan_to_num(views, nan=0, posinf=max_finite)

print("Original views:", views)
print("Updated views:", result)
```

```python
# Task 5
import numpy as np

numbers = np.array([11, 24, 37, 42, 55, 68, 73, 80])

even_indices = np.where(numbers % 2 == 0)[0]

print("Array:", numbers)
print("Indices of even numbers:", even_indices)
```
