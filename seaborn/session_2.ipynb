```python
# task 1
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Sample Swiggy orders dataset
df = pd.DataFrame({
    'Payment_Method': ['UPI', 'Card', 'Cash', 'UPI', 'Card',
                       'UPI', 'Cash', 'UPI', 'Card', 'Cash',
                       'UPI', 'Card', 'UPI', 'Cash', 'UPI']
})

sns.countplot(data=df, x='Payment_Method')

plt.title("Number of Orders by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Number of Orders")
plt.show()
```

```python
#  task 2
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Sample Zomato-style dataset
df = pd.DataFrame({
    'Food_Category': ['Pizza', 'Burger', 'Biryani', 'South Indian',
                      'Pizza', 'Burger', 'Biryani', 'South Indian',
                      'Pizza', 'Burger', 'Biryani', 'South Indian'],
    'Delivery_Time': [30, 25, 40, 35, 32, 20, 45, 30, 28, 27, 38, 33]
})

sns.barplot(data=df, x='Food_Category', y='Delivery_Time',
            estimator='mean', errorbar=None)

plt.title("Average Delivery Time by Food Category")
plt.xlabel("Food Category")
plt.ylabel("Average Delivery Time (Minutes)")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
#  task 3 
sns.barplot(data=df, x='Food_Category', y='Delivery_Time',
            estimator='sum', errorbar=None)

plt.title("Total Delivery Time by Food Category")
plt.xlabel("Food Category")
plt.ylabel("Total Delivery Time (Minutes)")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

sns.countplot(data=df, x='Food_Category')

plt.title("Total Orders by Food Category")
plt.xlabel("Food Category")
plt.ylabel("Number of Orders")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
#  task 4 
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Sample BookMyShow-style movie dataset
df_movies = pd.DataFrame({
    'Genre': ['Action', 'Comedy', 'Drama', 'Romance',
              'Action', 'Comedy', 'Drama', 'Romance',
              'Action', 'Comedy', 'Drama', 'Romance'],
    'Tickets_Sold': [120, 90, 100, 80, 150, 110, 95, 85,
                     130, 105, 115, 75]
})

sns.barplot(data=df_movies, x='Genre', y='Tickets_Sold', ci=None)

plt.title("Movie Tickets Sold by Genre")
plt.xlabel("Movie Genre")
plt.ylabel("Tickets Sold")
plt.show()
```
