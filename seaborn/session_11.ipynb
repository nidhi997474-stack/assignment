```python
#  task 1 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'category': ['Pizza', 'Burger', 'Biryani', 'Momos', 'Ice Cream'],
    'orders': [120, 150, 100, 90, 80]
})

plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='category', y='orders', color='steelblue')

plt.title("Orders per Food Category")
plt.xlabel("Food Category")
plt.ylabel("Number of Orders")
plt.tight_layout()

plt.savefig("food_orders_300dpi.png", dpi=300, bbox_inches="tight")
plt.show()

print("Saved: food_orders_300dpi.png")
```

```python
#  task  2
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
    'steps': [4500, 5200, 6100, 5800, 7000, 8500, 9000]
})

plt.figure(figsize=(9, 5))
sns.lineplot(data=df, x='day', y='steps', marker='o')

plt.title("Daily Step Counts for One Week")
plt.xlabel("Day")
plt.ylabel("Number of Steps")
plt.tight_layout()

plt.savefig("weekly_steps.svg", format="svg", bbox_inches="tight")
plt.show()

print("Saved: weekly_steps.svg")
```

```python
#  task  3
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'product': ['Mobile', 'Laptop', 'Headphones', 'Watch', 'Tablet'],
    'sales': [120, 95, 65, 80, 70]
})

sns.set_theme(style="darkgrid", palette="viridis")

plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='product', y='sales')

plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.tight_layout()

plt.savefig("product_sales_1200dpi.png", dpi=1200, bbox_inches="tight")
plt.show()

print("Saved: product_sales_1200dpi.png")
```

```python
# task 4 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

def apply_presentation_theme():
    sns.set_theme(
        style="whitegrid",
        palette="deep",
        font="sans-serif",
        font_scale=1.1
    )

apply_presentation_theme()

apps = ['WhatsApp', 'Instagram', 'YouTube', 'Chrome', 'Spotify']
usage = [35, 25, 20, 12, 8]

plt.figure(figsize=(8, 6))
plt.pie(
    usage,
    labels=apps,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Top 5 Most-Used Apps")
plt.tight_layout()

plt.savefig("top_apps_pie_chart.png", dpi=300, bbox_inches="tight")
plt.show()

print("Saved: top_apps_pie_chart.png")
```

```python
#  task  5 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'revenue': [100, 150, 200, 250, 300],
    'quantity': [10, 15, 20, 25, 30],
    'discount': [5, 10, 8, 15, 12],
    'profit': [20, 35, 45, 50, 65]
})

correlation = df.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap="coolwarm", vmin=-1, vmax=1)

plt.title("Revenue and Quantity Correlation Heatmap")
plt.tight_layout()

plt.savefig("correlation_heatmap.svg", format="svg", dpi=1200, bbox_inches="tight")
plt.show()

print("Saved: correlation_heatmap.svg")
```
