```python
#  task 1 
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(42)

df = pd.DataFrame({
    'delivery_time': np.random.randint(15, 60, 20),
    'order_value': np.random.randint(100, 1000, 20),
    'tip_amount': np.random.randint(10, 150, 20),
    'rating': np.random.uniform(2.5, 5.0, 20).round(1),
    'delivery_distance': np.random.uniform(1, 15, 20).round(2)
})

correlation_matrix = df.corr()

print("Food Delivery Dataset:")
print(df)

print("\nCorrelation Matrix:")
print(correlation_matrix)
```

```python
#  task 2 
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Food Delivery Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
#  task 3 
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='RdYlGn',
    vmin=-1,
    vmax=1,
    center=0,
    fmt='.2f'
)

plt.title("Correlation Heatmap with Diverging Palette")
plt.tight_layout()
plt.show()

#  to identify negative correlation 
corr_pairs = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
).stack()

if not corr_pairs.empty:
    print("Strongest negative correlation:")
    print(corr_pairs.idxmin(), "=", round(corr_pairs.min(), 2))
else:
    print("No feature pairs available.")
```

```python
#  task 4 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

phones = pd.DataFrame({
    'price': [9999, 12999, 15999, 19999, 24999,
              29999, 34999, 39999, 44999, 49999],
    'rating': [3.8, 4.0, 4.1, 4.2, 4.3,
               4.4, 4.5, 4.4, 4.6, 4.7],
    'RAM': [3, 4, 4, 6, 8, 8, 12, 12, 16, 16],
    'battery_capacity': [4000, 4500, 5000, 5000, 4800,
                         5000, 5500, 5000, 5200, 5500]
})

corr_phones = phones.corr()

sns.heatmap(
    corr_phones,
    annot=True,
    cmap='RdYlGn',
    vmin=-1,
    vmax=1,
    center=0,
    fmt='.2f'
)

plt.title("Mobile Phone Features Correlation Heatmap")
plt.tight_layout()
plt.show()
```
