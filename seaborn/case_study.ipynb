```python
import pandas as pd
import numpy as np

np.random.seed(42)

bank_df = pd.DataFrame({
    'age': np.random.randint(20, 70, 100),
    'job': np.random.choice(
        ['admin.', 'technician', 'services', 'management', 'retired'],
        100
    ),
    'marital': np.random.choice(
        ['married', 'single', 'divorced'], 100
    ),
    'education': np.random.choice(
        ['primary', 'secondary', 'tertiary', 'unknown'], 100
    ),
    'balance': np.random.randint(100, 10000, 100),
    'housing': np.random.choice(['yes', 'no'], 100),
    'loan': np.random.choice(['yes', 'no'], 100),
    'contact': np.random.choice(['cellular', 'telephone'], 100),
    'day': np.random.randint(1, 31, 100),
    'month': np.random.choice(
        ['jan', 'feb', 'mar', 'apr', 'may', 'jun'], 100
    ),
    'duration': np.random.randint(30, 1000, 100),
    'campaign': np.random.randint(1, 6, 100),
    'pdays': np.random.randint(-1, 500, 100),
    'previous': np.random.randint(0, 5, 100),
    'poutcome': np.random.choice(
        ['success', 'failure', 'unknown'], 100
    ),
    'y': np.random.choice(['yes', 'no'], 100)
})

bank_df.to_csv('bank.csv', index=False)

print("bank.csv created successfully!")
display(bank_df.head(10))
```

```python
bank_df = pd.read_csv("bank.csv")

print("First 10 rows:")
display(bank_df.head(10))

print("\nDataset Information:")
bank_df.info()

print("\nDataset Shape:")
print(bank_df.shape)

print("\nMissing Values:")
print(bank_df.isnull().sum())
```

```python
#  task 2 


import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(7, 5))

sns.countplot(data=bank_df, x='y', hue='y', legend=False)

plt.title("Term Deposit Subscription Count")
plt.xlabel("Subscribed to Term Deposit")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()
```

```python
#  task 3 

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.boxplot(data=bank_df, x='y', y='age')

plt.title("Customer Age vs Term Deposit Subscription")
plt.xlabel("Term Deposit Subscription")
plt.ylabel("Customer Age")

plt.tight_layout()
plt.show()
```

```python
#  task 4

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Calculate subscription percentage for each education level
education_summary = (
    bank_df.groupby('education')['y']
    .apply(lambda x: (x == 'yes').mean() * 100)
    .reset_index(name='subscription_percentage')
)

print(education_summary)

# Create bar plot
plt.figure(figsize=(8, 5))

sns.barplot(
    data=education_summary,
    x='education',
    y='subscription_percentage',
    color='skyblue'
)

plt.title('Term Deposit Subscription by Education Level')
plt.xlabel('Education Level')
plt.ylabel('Subscription Percentage (%)')

plt.tight_layout()
plt.show()
```

```python
#  task 5
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Calculate subscription percentage for each job
job_summary = (
    bank_df.groupby('job')['y']
    .apply(lambda x: (x == 'yes').mean() * 100)
    .reset_index(name='subscription_percentage')
    .sort_values('subscription_percentage', ascending=False)
)

print(job_summary)

# Create the bar plot
plt.figure(figsize=(10, 6))

sns.barplot(
    data=job_summary,
    x='job',
    y='subscription_percentage',
    color='teal'
)

plt.title('Term Deposit Subscription Rate by Job')
plt.xlabel('Job Category')
plt.ylabel('Subscription Rate (%)')
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```
