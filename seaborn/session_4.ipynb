```python
#  task 2
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Category": [
        "Fast Food", "Fast Food", "Fast Food", "Fast Food",
        "Fast Food", "Desserts", "Desserts", "Desserts",
        "Desserts", "Desserts"
    ],
    "Order_Amount": [
        150, 200, 250, 180, 300,
        100, 120, 150, 180, 220
    ]
})

sns.violinplot(data=df, x="Category", y="Order_Amount")

plt.title("Order Amounts by Restaurant Category")
plt.xlabel("Restaurant Category")
plt.ylabel("Order Amount (₹)")
plt.show()
```

```python
#  task 3 
import seaborn as sns
import matplotlib.pyplot as plt

ticket_prices = [
    500, 600, 700, 800, 900, 1000, 1200,
    1500, 1800, 2000, 2500, 3000, 15000
]

sns.boxplot(y=ticket_prices)

plt.title("IPL Match Ticket Price Distribution")
plt.ylabel("Ticket Price (₹)")
plt.show()

# Identify outliers using the IQR method
import numpy as np

Q1 = np.percentile(ticket_prices, 25)
Q3 = np.percentile(ticket_prices, 75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = [
    price for price in ticket_prices
    if price < lower_limit or price > upper_limit
]

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("Outliers:", outliers)
```

```python
#  task 4 
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

steps = [
    3000, 4500, 5000, 5500, 6000, 6500, 7000,
    7200, 7500, 7800, 8000, 8200, 8500, 9000,
    9500, 10000, 11000, 12000, 13000, 15000,
    4000, 6200, 7300, 8400, 9200, 10500, 11500,
    6800, 7600, 8800
]

df = pd.DataFrame({"Daily_Steps": steps})

sns.violinplot(data=df, y="Daily_Steps", inner="quartile")

plt.title("Distribution of Daily Step Counts")
plt.ylabel("Steps per Day")
plt.show()
```
