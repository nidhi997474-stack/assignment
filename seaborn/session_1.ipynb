```python
#  task 2 
import seaborn as sns
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.lineplot(x=days, y=temps, marker='o')

plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Temperature Trend")

plt.show()
```

```python
#  task 3  [A]
import seaborn as sns
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.set_theme(style="darkgrid")

ax = sns.lineplot(x=days, y=temps, marker='o')
ax.set_title("Weekly Temperature Trend")
ax.set_xlabel("Days")
ax.set_ylabel("Temperature (°C)")

plt.show()
```

```python
#  task 3 [B]
import seaborn as sns
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.set_theme(style="darkgrid")

g = sns.relplot(
    x=days, y=temps,
    kind="line", marker='o'
)

g.set_axis_labels("Days", "Temperature (°C)")
g.fig.suptitle("Weekly Temperature Trend")

plt.show()
```

```python
#  task 4 
import seaborn as sns
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

themes = ['darkgrid', 'whitegrid', 'ticks']

for theme in themes:
    sns.set_theme(style=theme)

    ax = sns.lineplot(x=days, y=temps, marker='o')
    ax.set_title("Weekly Temperature Trend")
    ax.set_xlabel("Days")
    ax.set_ylabel("Temperature (°C)")

    plt.savefig(f"temperature_{theme}.png", dpi=150)
    plt.show()
    plt.close()
```
