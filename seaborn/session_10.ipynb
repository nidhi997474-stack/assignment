```python
#  task 1 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'],
    'songs_streamed': [120, 150, 130, 180, 220, 300, 280]
})

sns.set_context('poster')

sns.barplot(data=df, x='day', y='songs_streamed', color='skyblue')

plt.title("Songs Streamed Per Day on Spotify")
plt.xlabel("Day")
plt.ylabel("Songs Streamed")
plt.show()

# Reset default context
sns.set_context('notebook')
```

```python
#  task 2 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'cuisine': ['Indian', 'Chinese', 'Italian', 'Indian', 'Chinese',
                'Italian', 'Indian', 'Chinese', 'Italian', 'Indian',
                'Chinese', 'Italian'],
    'orders': [120, 90, 75, 140, 100, 80, 110, 95, 85, 130, 105, 70]
})

sns.set_palette('colorblind')

sns.countplot(data=df, x='cuisine')

plt.title("Food Orders by Cuisine")
plt.xlabel("Cuisine Type")
plt.ylabel("Number of Orders")
plt.show()

sns.set_palette('deep')
```

```python
#  task 3 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'day': list(range(1, 15)),
    'steps': [4500, 5200, 6100, 5800, 7000, 7500, 6800,
              8000, 8500, 7800, 9000, 9500, 8700, 10000]
})

sns.set_palette('muted')
sns.set_context('notebook')

sns.lineplot(data=df, x='day', y='steps', marker='o')

plt.title("Daily Step Counts Over Two Weeks")
plt.xlabel("Day")
plt.ylabel("Number of Steps")
plt.tight_layout()
plt.show()

sns.set_palette('deep')
```

```python
#  task 4 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'category': ['Electronics', 'Clothing', 'Food', 'Books', 'Beauty'],
    'sales': [50000, 35000, 25000, 15000, 30000]
})

# Original styling
sns.set_palette('deep')
sns.set_context('talk')

sns.barplot(data=df, x='category', y='sales')
plt.title("Sales by Category - Deep Palette")
plt.tight_layout()
plt.show()

# Updated styling
sns.set_palette('colorblind')
sns.set_context('poster')

sns.barplot(data=df, x='category', y='sales')
plt.title("Sales by Category - Colorblind Palette")
plt.tight_layout()
plt.show()

# Reset styling
sns.set_palette('deep')
sns.set_context('notebook')
```

```python
#  task  5 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'product': ['Mobiles', 'Laptops', 'Headphones', 'Smartwatches', 'Tablets'],
    'sales': [120000, 95000, 65000, 80000, 70000]
})

sns.set(
    palette='deep',
    context='talk',
    font_scale=1.0,
    style='whitegrid'
)

plt.figure(figsize=(10, 6))

sns.barplot(data=df, x='product', y='sales')

plt.title("Flipkart-Style Product Sales Dashboard")
plt.xlabel("Product Category")
plt.ylabel("Sales (₹)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

# Reset to standard styling
sns.set_theme(context='notebook', style='darkgrid', palette='deep')
```
