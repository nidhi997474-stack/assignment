```python
# task 1
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Sample Swiggy delivery time dataset (30 values)
delivery_time = [
    20, 25, 30, 35, 28, 32, 40, 22, 26, 31,
    29, 35, 38, 24, 27, 33, 36, 30, 25, 42,
    34, 28, 31, 39, 23, 26, 37, 32, 29, 45
]

df = pd.DataFrame({'Delivery_Time': delivery_time})

sns.histplot(data=df, x='Delivery_Time', bins=10, kde=False)

plt.title("Distribution of Delivery Times")
plt.xlabel("Delivery Time (Minutes)")
plt.ylabel("Frequency")
plt.show()
```

```python
#  task 2 
import seaborn as sns
import matplotlib.pyplot as plt

# Daily step counts for one month
steps = [
    4500, 6200, 7000, 8000, 5500, 9000, 7500,
    6800, 7200, 8500, 10000, 6500, 7800, 9200,
    5000, 6000, 8200, 7300, 9500, 11000, 8700,
    6900, 7600, 8800, 5400, 6700, 9800, 10500,
    8100, 7200
]

sns.kdeplot(x=steps, fill=True)

plt.title("Distribution of Daily Step Counts")
plt.xlabel("Steps per Day")
plt.ylabel("Probability Density")
plt.show()
```

```python
#  task 3 
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Monthly spending amounts in rupees
spending = [
    250, 400, 550, 700, 350, 900, 1200, 650, 800, 450,
    1500, 1100, 600, 750, 950, 1300, 500, 850, 1000, 700,
    1600, 1400, 300, 550, 900, 1250, 800, 650, 1800, 2000
]

df_spending = pd.DataFrame({'Monthly_Spending': spending})

sns.histplot(
    data=df_spending,
    x='Monthly_Spending',
    bins=10,
    stat='density',
    color='skyblue'
)

sns.kdeplot(
    data=df_spending,
    x='Monthly_Spending',
    color='red',
    linewidth=2
)

plt.title("Distribution of Monthly Spending on Zomato Orders")
plt.xlabel("Monthly Spending (₹)")
plt.ylabel("Density")
plt.show()
```

```python
#  task 4 
import seaborn as sns
import matplotlib.pyplot as plt

# Sample IPL match scores
runs = [
    120, 145, 160, 175, 190, 135, 155, 180, 200, 165,
    150, 170, 185, 210, 125, 140, 195, 220, 155, 175,
    230, 160, 185, 205, 145, 190, 215, 170, 180, 250
]

# Compare different bin sizes
for bins in [5, 10, 20]:
    sns.histplot(x=runs, bins=bins, color='green')

    plt.title(f"IPL Run Distribution ({bins} Bins)")
    plt.xlabel("Runs Scored")
    plt.ylabel("Frequency")
    plt.show()
    plt.close()
```
