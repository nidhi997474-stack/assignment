```python
#  task 1 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'likes': [100, 200, 300, 400, 500, 600, 700, 800, 900, 1000],
    'comments': [10, 25, 30, 45, 50, 65, 70, 85, 90, 105]
})

sns.regplot(data=df, x='likes', y='comments')

plt.title("Instagram Likes vs Comments")
plt.xlabel("Number of Likes")
plt.ylabel("Number of Comments")
plt.show()
```

```python
#  task  3

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm

df = pd.DataFrame({
    'songs': [5, 8, 10, 12, 15, 18, 20, 22, 25, 28, 30, 35],
    'duration': [15, 24, 30, 38, 45, 55, 60, 68, 75, 90, 95, 110]
})

# Calculate LOWESS curve
lowess = sm.nonparametric.lowess(
    df['duration'],
    df['songs'],
    frac=0.6
)

# Plot data points
sns.scatterplot(data=df, x='songs', y='duration', color='blue')

# Plot LOWESS smoothing curve
plt.plot(lowess[:, 0], lowess[:, 1], color='red')

plt.title("Playlist Songs vs Total Duration")
plt.xlabel("Number of Songs")
plt.ylabel("Duration (Minutes)")
plt.show()
```

```python
#  task 5 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'price': [500, 800, 1000, 1500, 2000, 2500, 3000, 3500, 4000, 5000],
    'reviews': [20, 35, 50, 65, 80, 100, 120, 140, 160, 200]
})

# Customize plot aesthetics only
sns.regplot(
    data=df,
    x='price',
    y='reviews',
    scatter_kws={'color': 'green', 's': 60},
    line_kws={'color': 'red'}
)

plt.title("Flipkart Product Price vs Number of Reviews")
plt.xlabel("Product Price (₹)")
plt.ylabel("Number of Reviews")
plt.tight_layout()
plt.show()
```
