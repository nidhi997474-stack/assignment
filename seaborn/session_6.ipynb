```python
#  task 1
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load CSV file
df = pd.read_csv("flipkart_products.csv")

# Scatter jointplot
sns.jointplot(
    data=df,
    x="rating",
    y="price",
    kind="scatter"
)

plt.show()
```

```python
#  task 3 


import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "likes": [100, 150, 200, 250, 300, 350, 400, 450, 500, 600,
              180, 220, 280, 320, 380, 420, 480, 550, 650, 700],
    "comments": [12, 18, 16, 30, 25, 40, 35, 50, 45, 62,
                 20, 24, 32, 29, 42, 38, 52, 48, 68, 75]
})

sns.kdeplot(
    data=df,
    x="likes",
    y="comments",
    fill=False,
    levels=5
)

plt.title("Instagram Likes vs Comments")
plt.xlabel("Number of Likes")
plt.ylabel("Number of Comments")
plt.show()
```

```python
#  task  4
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Zomato restaurant review data
df = pd.DataFrame({
    "rating": [3.2, 3.5, 3.8, 4.0, 4.1, 4.2, 4.3, 4.5, 4.6, 4.8,
               3.7, 4.0, 4.2, 4.4, 4.5, 3.9, 4.1, 4.3, 4.6, 4.7],
    "reviews": [50, 80, 120, 150, 180, 200, 250, 300, 350, 400,
                100, 140, 190, 220, 280, 130, 170, 240, 320, 380]
})

# Scatter plot
sns.jointplot(data=df, x="rating", y="reviews", kind="scatter")
plt.show()

# Hexbin plot
sns.jointplot(data=df, x="rating", y="reviews", kind="hex")
plt.show()

# KDE plot
sns.jointplot(data=df, x="rating", y="reviews", kind="kde", fill=True)
plt.show()
```
