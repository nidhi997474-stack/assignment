```python
#  task 1 
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
active_users = [1200, 1500, 1350, 1800, 2100, 2500, 2800]

plt.figure(figsize=(10, 4))
plt.plot(days, active_users, marker='o')

plt.title('Daily Active Users - Music Streaming App')
plt.xlabel('Days')
plt.ylabel('Number of Active Users')

plt.show()
```

```python
#  task 2 
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
zomato_orders = [200, 250, 230, 300, 350, 400, 450]
swiggy_orders = [180, 220, 260, 280, 320, 380, 420]

plt.plot(days, zomato_orders, linestyle='--',
         marker='o', label='Zomato')

plt.plot(days, swiggy_orders, linestyle=':',
         marker='s', label='Swiggy')

plt.title('Weekly Food Orders: Zomato vs Swiggy')
plt.xlabel('Days')
plt.ylabel('Number of Orders')
plt.legend()

plt.show()
```

```python
#  task 3 
import matplotlib.pyplot as plt

days = range(1, 11)
sales = [120, 150, 130, 180, 200, 170, 220, 250, 230, 280]

plt.style.use('ggplot')

plt.plot(days, sales, color='orange', marker='o')

plt.title('Flipkart Daily Sales for 10 Days')
plt.xlabel('Day')
plt.ylabel('Sales')

plt.show()
```

```python
#  task 4 
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
posts = [12, 18, 15, 22, 25, 30, 28]

instagram_colors = [
    '#833AB4', '#C13584', '#E1306C',
    '#FD1D1D', '#F77737', '#C13584', '#833AB4'
]

plt.bar(days, posts, color=instagram_colors)

plt.title('Daily Posts Uploaded on Instagram')
plt.xlabel('Days')
plt.ylabel('Number of Posts')

plt.show()
```

```python
#  task 5 
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
sales = [100, 140, 120, 180, 160, 220, 250]

plt.style.use('dark_background')

plt.plot(days, sales, color='cyan',
         marker='o', linewidth=2, label='Sales')

plt.title('Weekly Sales - Dark Theme')
plt.xlabel('Days')
plt.ylabel('Sales')
plt.legend()

plt.show()
```
