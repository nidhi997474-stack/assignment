```python
#  task 1
import matplotlib.pyplot as plt
import numpy as np

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
zomato = [200, 250, 230, 300, 350, 400, 450]
swiggy = [180, 220, 260, 280, 320, 380, 420]

plt.bar(days, zomato, color='red', label='Zomato')
plt.bar(days, swiggy, bottom=zomato, color='orange', label='Swiggy')

plt.title('Weekly Food Orders')
plt.xlabel('Days')
plt.ylabel('Number of Orders')
plt.legend()

plt.show()
```

```python
#  task 2 
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
spotify = [120, 150, 140, 180, 200, 220, 250]
youtube = [200, 220, 210, 250, 270, 300, 320]

plt.fill_between(days, spotify, color='green',
                 alpha=0.5, label='Spotify')

plt.fill_between(days, youtube, color='red',
                 alpha=0.5, label='YouTube')

plt.title('Daily Active Users: Spotify vs YouTube')
plt.xlabel('Days')
plt.ylabel('Active Users (in thousands)')
plt.legend()

plt.show()
```

```python
#  task 3 
import matplotlib.pyplot as plt

days = range(1, 11)
delivery_time = [30, 35, 32, 40, 38, 34, 42, 36, 39, 33]
std_deviation = [3, 4, 2, 5, 3, 4, 2, 3, 4, 2]

plt.errorbar(days, delivery_time, yerr=std_deviation,
             fmt='o-', color='blue', capsize=5)

plt.title('Flipkart Average Delivery Time')
plt.xlabel('Day')
plt.ylabel('Delivery Time (minutes)')

plt.show()
```

```python
# task 4
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime, timedelta

start_date = datetime(2026, 10, 1)
dates = [start_date + timedelta(days=i) for i in range(14)]

balance = [1500, 1450, 1600, 1350, 1400, 1550, 1300,
           1250, 1450, 1500, 1650, 1550, 1700, 1800]

plt.plot(dates, balance, marker='o', color='purple')

plt.title('Paytm Wallet Balance - Last 14 Days')
plt.xlabel('Date')
plt.ylabel('Wallet Balance (₹)')

plt.gca().xaxis.set_major_formatter(
    mdates.DateFormatter('%d %b')
)
plt.gcf().autofmt_xdate()

plt.show()
```

```python
#  task 5 
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
spotify = [120, 150, 140, 180, 200, 220, 250]
youtube = [200, 220, 210, 250, 270, 300, 320]

fig, ax = plt.subplots(figsize=(10, 5))

ax.fill_between(days, spotify, color='#1DB954',
                alpha=0.6, label='Spotify')

ax.fill_between(days, youtube, color='#FF0000',
                alpha=0.4, label='YouTube')

ax.set_title('Social Media Analytics Dashboard')
ax.set_xlabel('Day of the Week')
ax.set_ylabel('Daily Active Users (in thousands)')
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend()

plt.show()
```
