```python
# task 1 
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [4000, 5500, 6000, 7500, 6500, 9000, 8000]

plt.plot(days, steps, color="green", linewidth=2)

plt.show()
```

```python
# task 2
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
users = [120, 135, 150, 145, 170, 190]  # Sample data in millions

plt.plot(
    months,
    users,
    marker="o",
    markerfacecolor="red"
)

plt.xlabel("Month")
plt.ylabel("Monthly Active Users (millions)")
plt.title("Monthly Active Users of a Music App")

plt.show()
```

```python
#  task 3 
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

zomato_orders = [8, 10, 7, 12, 9]
swiggy_orders = [6, 9, 11, 8, 13]

plt.plot(
    months,
    zomato_orders,
    color="red",
    linestyle="-",
    marker="o",
    label="Zomato"
)

plt.plot(
    months,
    swiggy_orders,
    color="orange",
    linestyle="--",
    marker="s",
    label="Swiggy"
)

plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.title("Monthly Orders: Zomato vs Swiggy")
plt.legend()

plt.show()
```

```python
#  task 4
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
data_usage = [12, 15, 10, 18, 16, 20]

plt.plot(months, data_usage, marker="o", color="blue")

plt.title("Monthly Data Usage Trend")
plt.xlabel("Month")
plt.ylabel("Data Used (GB)")

plt.show()
```
