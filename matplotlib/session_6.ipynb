```python
#  task 2 
import matplotlib.pyplot as plt

friends = [
    "Asha", "Ravi", "Priya", "Karan", "Neha", "Arjun",
    "Meera", "Dev", "Isha", "Rahul", "Anaya", "Vikram"
]

orders = [2, 5, 3, 8, 4, 6, 7, 2, 9, 5, 4, 10]
spending = [400, 1200, 650, 1800, 900, 1500, 1600, 350, 2200, 1100, 850, 2500]

plt.scatter(orders, spending, color="teal")

for i, friend in enumerate(friends):
    plt.annotate(
        friend,
        (orders[i], spending[i]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.xlabel("Number of Zomato Orders")
plt.ylabel("Amount Spent (₹)")
plt.title("Zomato Orders vs Spending")

plt.show()
```

```python
#  task 3 
import matplotlib.pyplot as plt
import numpy as np

orders = [2, 5, 3, 8, 4, 6, 7, 2, 9, 5, 4, 10]
spending = [400, 1200, 650, 1800, 900, 1500, 1600, 350, 2200, 1100, 850, 2500]

# Calculate slope and intercept
slope, intercept = np.polyfit(orders, spending, 1)

# Calculate predicted spending
trend_line = np.polyval([slope, intercept], orders)

plt.scatter(orders, spending, color="teal", label="Actual data")
plt.plot(orders, trend_line, color="red", label="Trend line")

plt.xlabel("Number of Zomato Orders")
plt.ylabel("Amount Spent (₹)")
plt.title("Zomato Orders vs Spending with Trend Line")
plt.legend()
plt.show()
```

```python
#  task 4 
import matplotlib.pyplot as plt
import numpy as np

youtube_hours = [
    1, 2, 3, 1.5, 4, 2.5, 5, 3.5, 2, 4.5,
    6, 3, 5.5, 1.8, 4.2
]

spotify_hours = [
    2, 1.5, 3, 2.2, 2.5, 3.5, 4, 3, 2.8, 4.2,
    3.8, 2.6, 4.5, 1.9, 3.6
]

# Fit a straight regression line
slope, intercept = np.polyfit(youtube_hours, spotify_hours, 1)

# Generate points for a smooth trend line
x_line = np.linspace(min(youtube_hours), max(youtube_hours), 100)
y_line = slope * x_line + intercept

plt.scatter(
    youtube_hours,
    spotify_hours,
    color="blue",
    label="Users"
)

plt.plot(x_line, y_line, color="red", label="Regression line")

equation = f"y = {slope:.2f}x + {intercept:.2f}"

plt.text(
    0.05, 0.95,
    equation,
    transform=plt.gca().transAxes,
    verticalalignment="top"
)

plt.xlabel("YouTube Watch Hours")
plt.ylabel("Spotify Listening Hours")
plt.title("YouTube vs Spotify Usage")
plt.legend()
plt.show()

print("Regression equation:", equation)
```

```python
#  task 5 
import matplotlib.pyplot as plt
import numpy as np

# Sample data for 10 users
orders = [1, 3, 2, 5, 4, 6, 2, 7, 5, 8]
amount_spent = [500, 1800, 1200, 3000, 2400, 3500, 1000, 4200, 2800, 4800]

# Calculate regression line
slope, intercept = np.polyfit(orders, amount_spent, 1)

# Generate smooth line values
x_line = np.linspace(min(orders), max(orders), 100)
y_line = slope * x_line + intercept

plt.scatter(
    orders,
    amount_spent,
    color="green",
    label="Users"
)

plt.plot(
    x_line,
    y_line,
    color="red",
    label="Regression line"
)

plt.xlabel("Number of Flipkart Orders")
plt.ylabel("Total Amount Spent (₹)")
plt.title("Flipkart Orders vs Total Spending")
plt.legend()
plt.show()

print(f"Regression equation: y = {slope:.2f}x + {intercept:.2f}")
```
