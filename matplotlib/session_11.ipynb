```python
#  task 1
import matplotlib.pyplot as plt

restaurants = ['Restaurant A', 'Restaurant B',
               'Restaurant C', 'Restaurant D',
               'Restaurant E']

orders = [120, 150, 100, 180, 140]

plt.figure(figsize=(8, 5))
plt.bar(restaurants, orders, color='orange')

plt.title('Orders from Zomato Restaurants')
plt.xlabel('Restaurants')
plt.ylabel('Number of Orders')
plt.xticks(rotation=20)

plt.savefig('zomato_orders.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
#  task 2 
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
steps = [4500, 6000, 5200, 7500, 8000, 10000, 9000]

plt.figure(figsize=(8, 5))
plt.plot(days, steps, marker='o', color='green')

plt.title('Daily Step Count')
plt.xlabel('Days')
plt.ylabel('Number of Steps')
plt.grid(True)

plt.savefig('daily_steps.svg', bbox_inches='tight')
plt.show()
```

```python
#  task 3
import matplotlib.pyplot as plt

apps = ['Instagram', 'YouTube', 'WhatsApp', 'Spotify']
hours = [3, 2, 1.5, 1.5]

plt.figure(figsize=(7, 7))

plt.pie(
    hours,
    labels=apps,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Daily Time Spent on Apps')

plt.savefig('app_usage.pdf', bbox_inches='tight')
plt.show()
```

```python
#  task 5
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [100, 150, 130, 180, 220, 250]

plt.figure(figsize=(10, 6))

plt.plot(months, sales, marker='o', linewidth=2)

plt.title('Monthly Sales Report')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)

# Save the chart as a PDF
plt.savefig('monthly_sales.pdf', dpi=300, bbox_inches='tight')

plt.show()
```
