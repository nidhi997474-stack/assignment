```python
#  task 1 
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [4000, 5500, 6000, 7500, 6500, 9000, 8000]
water = [2.0, 2.5, 2.2, 3.0, 2.5, 3.2, 2.8]

plt.figure(figsize=(9, 7))

# First subplot: Daily steps
plt.subplot(2, 1, 1)
plt.plot(days, steps, marker="o", color="blue")
plt.title("Daily Step Count")
plt.ylabel("Steps")
plt.grid(True)

# Second subplot: Water intake
plt.subplot(2, 1, 2)
plt.plot(days, water, marker="o", color="green")
plt.title("Daily Water Intake")
plt.xlabel("Day")
plt.ylabel("Water (litres)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
#  task 2 
import matplotlib.pyplot as plt

quarters = ["Q1", "Q2", "Q3", "Q4"]
posts = [12, 18, 15, 24]

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Line chart
axes[0, 0].plot(quarters, posts, marker="o", color="blue")
axes[0, 0].set_title("Line Chart")
axes[0, 0].set_ylabel("Number of Posts")

# Bar chart
axes[0, 1].bar(quarters, posts, color="orange")
axes[0, 1].set_title("Bar Chart")
axes[0, 1].set_ylabel("Number of Posts")

# Scatter plot
axes[1, 0].scatter(quarters, posts, color="green")
axes[1, 0].set_title("Scatter Plot")
axes[1, 0].set_ylabel("Number of Posts")

# Pie chart
axes[1, 1].pie(posts, labels=quarters, autopct="%1.1f%%")
axes[1, 1].set_title("Pie Chart")

fig.suptitle("Instagram Posts by Quarter")
plt.tight_layout()
plt.show()
```

```python
# task 3 
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

zomato = [8, 10, 7, 12, 9, 14]
swiggy = [6, 9, 11, 8, 13, 10]
dominos = [4, 5, 6, 7, 5, 8]

fig, axes = plt.subplots(
    1, 3,
    figsize=(14, 4),
    sharey=True
)

axes[0].plot(months, zomato, marker="o", color="red")
axes[0].set_title("Zomato Orders")
axes[0].set_ylabel("Number of Orders")

axes[1].plot(months, swiggy, marker="o", color="orange")
axes[1].set_title("Swiggy Orders")

axes[2].plot(months, dominos, marker="o", color="blue")
axes[2].set_title("Domino's Orders")

for ax in axes:
    ax.set_xlabel("Month")
    ax.grid(True)

fig.suptitle("Monthly Food Delivery Orders")
plt.tight_layout()
plt.show()
```

```python
#  task 4 
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

youtube = [8, 10, 12, 9, 14, 16]
instagram = [5, 7, 8, 6, 9, 11]
spotify = [3, 4, 3.5, 5, 4.5, 6]
whatsapp = [2, 2.5, 3, 2.8, 3.5, 4]

fig, axes = plt.subplots(
    2, 2,
    figsize=(10, 7),
    sharex=True
)

axes[0, 0].plot(months, youtube, marker="o")
axes[0, 0].set_title("YouTube")
axes[0, 0].set_ylabel("Data Used (GB)")

axes[0, 1].plot(months, instagram, marker="o", color="orange")
axes[0, 1].set_title("Instagram")
axes[0, 1].set_ylabel("Data Used (GB)")

axes[1, 0].plot(months, spotify, marker="o", color="green")
axes[1, 0].set_title("Spotify")
axes[1, 0].set_xlabel("Month")
axes[1, 0].set_ylabel("Data Used (GB)")

axes[1, 1].plot(months, whatsapp, marker="o", color="purple")
axes[1, 1].set_title("WhatsApp")
axes[1, 1].set_xlabel("Month")
axes[1, 1].set_ylabel("Data Used (GB)")

fig.suptitle("Monthly Mobile Data Usage by App")
plt.tight_layout()
plt.show()
```
