```python
#  task 1 
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

users = [1200, 1500, 1400, 1800, 2000, 2200,
         2100, 2500, 2400, 2800, 3000, 3500]

plt.figure(figsize=(10, 5))
plt.plot(months, users, marker='o', color='green')

# Find the month with the highest users
max_index = users.index(max(users))

plt.annotate(
    'Highest Users',
    xy=(months[max_index], users[max_index]),
    xytext=(months[max_index - 2], users[max_index] - 500),
    arrowprops=dict(arrowstyle='->', color='red'),
    color='red'
)

plt.title('Monthly Active Users - Music Streaming App')
plt.xlabel('Month')
plt.ylabel('Number of Active Users')
plt.grid(True)

plt.show()
```

```python
#  task 2
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

orders = [450, 380, 500, 420, 600, 550,
          480, 700, 650, 580, 750, 800]

plt.figure(figsize=(10, 5))
plt.bar(months, orders, color='skyblue')

# Find the month with the lowest orders
min_index = orders.index(min(orders))

plt.annotate(
    'Lowest Orders',
    xy=(min_index, orders[min_index]),
    xytext=(min_index + 1, orders[min_index] + 100),
    arrowprops=dict(arrowstyle='->', color='red'),
    color='red'
)

plt.title('Monthly Food Delivery Orders')
plt.xlabel('Month')
plt.ylabel('Number of Orders')

plt.show()
```

```python
#  task 3 
import matplotlib.pyplot as plt

methods = ['UPI', 'Card', 'Wallet', 'Cash']
percentages = [50, 25, 15, 10]

plt.figure(figsize=(8, 6))

plt.pie(
    percentages,
    labels=methods,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Payment Methods Used on E-commerce App')

plt.legend(
    methods,
    loc='lower right',
    title='Payment Methods'
)

plt.show()
```

```python
#  task 4 
import matplotlib.pyplot as plt

ratings = [7.5, 8.2, 6.8, 9.1, 8.7, 7.9, 9.5, 8.0]
reviews = [1200, 2500, 900, 3000, 1800, 1500, 4000, 2200]

plt.figure(figsize=(9, 5))
plt.scatter(reviews, ratings, color='blue', s=80)

# Find the movie with the highest rating
max_index = ratings.index(max(ratings))

plt.annotate(
    'Highest Rating: 9.5',
    xy=(reviews[max_index], ratings[max_index]),
    xytext=(reviews[max_index] - 1800, ratings[max_index] - 0.7),
    bbox=dict(boxstyle='round,pad=0.5',
              fc='yellow', ec='black'),
    arrowprops=dict(arrowstyle='->', color='red')
)

plt.title('Movie Ratings vs Number of Reviews')
plt.xlabel('Number of Reviews')
plt.ylabel('Movie Rating')
plt.grid(True)

plt.show()
```
