```python
#  task 1
import matplotlib.pyplot as plt

# Food delivery apps
apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]

# Sample daily order counts
orders = [120, 100, 80, 60, 90]

# Custom colors for each bar
colors = ["red", "orange", "blue", "green", "purple"]

# Create vertical bar chart
plt.bar(apps, orders, color=colors)

plt.xlabel("Food Delivery Apps")
plt.ylabel("Number of Daily Orders")
plt.title("Daily Orders by Food Delivery App")

plt.show()
```

```python
#  task 3 
import matplotlib.pyplot as plt

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]
orders = [120, 100, 80, 60, 90]
colors = ["red", "orange", "blue", "green", "purple"]

# Create vertical bar chart
bars = plt.bar(apps, orders, color=colors)

# Add order counts above each bar
for bar in bars:
    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        str(height),
        ha="center",
        va="bottom"
    )

plt.xlabel("Food Delivery Apps")
plt.ylabel("Number of Daily Orders")
plt.title("Daily Orders by Food Delivery App")

plt.show()
```

```python
# task 4 
import matplotlib.pyplot as plt

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]
orders = [120, 100, 80, 60, 90]
colors = ["red", "orange", "blue", "green", "purple"]

# Create thinner bars
bars = plt.bar(
    apps,
    orders,
    color=colors,
    width=0.5
)

# Add labels above the bars
for bar in bars:
    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        str(height),
        ha="center",
        va="bottom"
    )

plt.xlabel("Food Delivery Apps")
plt.ylabel("Number of Daily Orders")
plt.title("Daily Orders by Food Delivery App")

plt.show()
```
