```python
#  task 2
import matplotlib.pyplot as plt

scores = [
    145, 178, 162, 190, 155, 201, 175, 168, 184, 152,
    195, 172, 160, 188, 210, 149, 177, 165, 199, 182,
    157, 220, 173, 186, 143, 193, 169, 205, 158, 180,
    196, 147, 174, 189, 163, 212, 176, 154, 191, 167,
    185, 202, 159, 179, 198, 171, 150, 187, 166, 194
]

# Histogram with 5 bins
plt.hist(scores, bins=5, color="orange", edgecolor="black")
plt.title("IPL Scores — 5 Bins")
plt.xlabel("Runs")
plt.ylabel("Frequency")
plt.show()

# Histogram with 10 bins
plt.hist(scores, bins=10, color="green", edgecolor="black")
plt.title("IPL Scores — 10 Bins")
plt.xlabel("Runs")
plt.ylabel("Frequency")
plt.show()

# Histogram with 15 bins
plt.hist(scores, bins=15, color="purple", edgecolor="black")
plt.title("IPL Scores — 15 Bins")
plt.xlabel("Runs")
plt.ylabel("Frequency")
plt.show()
```

```python
#  task 3 
import matplotlib.pyplot as plt

steps = [
    4200, 5100, 6800, 7500, 6200, 8100, 9000,
    7200, 5600, 6400, 7800, 8500, 9200, 4800,
    5300, 6900, 7400, 8800, 9600, 6100, 7000,
    8200, 5900, 7700, 8600, 9400, 6500, 7300,
    8000, 5500
]

# Frequency histogram
plt.hist(steps, bins=8, color="steelblue", edgecolor="black")
plt.title("Daily Steps — Frequency")
plt.xlabel("Number of Steps")
plt.ylabel("Frequency (days)")
plt.show()

# Density histogram
plt.hist(
    steps,
    bins=8,
    density=True,
    color="coral",
    edgecolor="black"
)
plt.title("Daily Steps — Density")
plt.xlabel("Number of Steps")
plt.ylabel("Probability Density")
plt.show()
```

```python
#  task 4 
import matplotlib.pyplot as plt
import random

# Generate 50 random song durations
durations = [
    random.randint(120, 300)
    for _ in range(50)
]

plt.hist(
    durations,
    bins=10,
    color="purple",
    edgecolor="black"
)

plt.title("Spotify Song Duration Distribution")
plt.xlabel("Song Duration (seconds)")
plt.ylabel("Frequency")

plt.show()
```

```python
#  task 5 
import matplotlib.pyplot as plt

prices = [
    150, 250, 450, 500, 600, 750, 950,
    1100, 1350, 1800, 2000, 2200, 2800,
    3500, 4200, 4800
]

# Define bin edges
bins = [100, 500, 1000, 2000, 5000]

plt.hist(prices, bins=bins, color="gold", edgecolor="black")

plt.title("Flipkart Product Price Distribution")
plt.xlabel("Product Price (₹)")
plt.ylabel("Number of Products")

plt.xticks(bins)
plt.show()
```
