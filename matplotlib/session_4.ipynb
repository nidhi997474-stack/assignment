```python
#  task 1
import matplotlib.pyplot as plt

apps = ["Instagram", "YouTube", "WhatsApp", "Zomato", "Spotify"]
hours = [3, 2.5, 2, 1, 1.5]

plt.pie(
    hours,
    labels=apps,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Daily App Usage Distribution")
plt.show()
```

```python
#  task 2 
import matplotlib.pyplot as plt

categories = [
    "Food Delivery",
    "Shopping",
    "Entertainment",
    "UPI Payments"
]

spending = [2500, 4000, 1500, 2000]
explode = (0, 0.15, 0, 0)

plt.pie(
    spending,
    labels=categories,
    autopct="%1.1f%%",
    explode=explode,
    startangle=90
)

plt.title("Monthly Online Spending")
plt.show()
```

```python
#  task 3 
import matplotlib.pyplot as plt

categories = [
    "Food Delivery",
    "Shopping",
    "Entertainment",
    "UPI Payments"
]

spending = [2500, 4000, 1500, 2000]

fig, ax = plt.subplots()

ax.pie(
    spending,
    labels=categories,
    autopct="%1.1f%%",
    startangle=90
)

# Add a white circle at the center
centre_circle = plt.Circle(
    (0, 0),
    0.60,
    color="white"
)

ax.add_artist(centre_circle)

ax.set_title("Monthly Online Spending Donut Chart")
ax.axis("equal")

plt.show()
```

```python
#  task 4 
import matplotlib.pyplot as plt

teams = ["CSK", "MI", "RCB", "KKR"]

# Sample follower counts in millions
followers = [16, 14, 13, 8]

fig, ax = plt.subplots()

ax.pie(
    followers,
    labels=teams,
    autopct="%1.1f%%",
    startangle=90
)

# Create the donut hole
centre_circle = plt.Circle(
    (0, 0),
    0.60,
    color="white"
)

ax.add_artist(centre_circle)

ax.set_title("IPL Teams Instagram Followers Share")
ax.axis("equal")

plt.show()
```
